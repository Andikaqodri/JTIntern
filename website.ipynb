```python
import pandas as pd
```

```python
df = pd.read_excel("PBL_DATA_preprocessed.xlsx")
df.head()
```

```python
df['Minat_Utama'] = df['Minat_Bidang'].apply(lambda x: x.split(',')[0])
```

```python
prior = df['Minat_Utama'].value_counts(normalize=True)
print(prior)
```

```python
likelihood = df.groupby('Minat_Utama')['IPK'].mean()
print(likelihood)
```

```python
ipk_input = 3.8

hasil = {}

for kelas in prior.index:
    P_kelas = prior[kelas]
    mean_ipk = likelihood[kelas]

    # pendekatan sederhana (semakin dekat semakin besar peluang)
    P_x_kelas = 1 / (1 + abs(ipk_input - mean_ipk))

    hasil[kelas] = P_x_kelas * P_kelas

# hasil probabilitas
print(hasil)

# ambil yang terbesar
prediksi = max(hasil, key=hasil.get)
print("Prediksi Minat:", prediksi)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Minat_Utama'] = le.fit_transform(df['Minat_Utama'])

X = df[['IPK']]
y = df['Minat_Utama']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = GaussianNB()
model.fit(X_train, y_train)

print(model.score(X_test, y_test))
```
